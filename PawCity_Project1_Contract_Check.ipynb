## Preserve your previous comparison results

If your previous PawCity project saved `reports/sample_comparisons.json`, the cell below copies that checkpoint from a sibling project folder on your Desktop into this project. It does not call the LLM or overwrite an existing checkpoint.


```python
from pathlib import Path
import shutil
destination = ROOT / "reports" / "sample_comparisons.json"
desktop = Path.home() / "Desktop"
candidates = [p for p in desktop.glob("PawCity_AI_Complete*/reports/sample_comparisons.json") if p.resolve() != destination.resolve()]
candidates.sort(key=lambda p: p.stat().st_mtime, reverse=True)
if destination.exists():
    print("Checkpoint already exists; no file changed:", destination)
elif candidates:
    print("Found previous checkpoints:", *[str(p) for p in candidates], sep="\n - ")
    source = candidates[0]
    destination.parent.mkdir(parents=True, exist_ok=True)
    shutil.copy2(source, destination)
    print("Copied latest saved checkpoint from:", source)
else:
    print("No prior checkpoint found automatically. Keep your previous reports/ directory if applicable.")
```
