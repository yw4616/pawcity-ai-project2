## Verify model-specific comparison provenance

The final Project 1 API contract is unchanged. However, GPT responses cannot be reused for the Gemini-only baseline comparison. This notebook checks the current comparison model without copying previous files or making model requests.


```python
import json
checkpoint = ROOT / "reports" / "sample_comparisons.json"
if checkpoint.exists():
    saved = json.loads(checkpoint.read_text(encoding="utf-8"))
    print("Checkpoint model:", saved.get("model", "unknown"))
    print("Completed agent results:", sum(bool(x.get("agent_answer")) for x in saved.get("comparisons", [])))
    if not str(saved.get("model", "")).startswith("gemini-"):
        print("WARNING: Archive this non-Gemini checkpoint before Gemini comparisons.")
else:
    print("No comparison checkpoint found. Gemini model results must be generated live.")
```
